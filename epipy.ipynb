# begining stock market problem.


```python
import random
import matplotlib.pyplot as plt
''' there is some kind of problem i needed to solve given 3 arrays of stock market

 all non negetive floating point number,

l = lowest
h = highes 
s = starting
of a stock market
'''


#taking n days in consideration 

n = 10

#defining the variables
l =[]
h =[]
s =[]

#the max value that can be reached by the hypothetical stock market
max = 1000.0

for i in range(n):
    s.append(random.uniform(0.0,max))
    l.append(random.uniform(0.0,s[i]))
    h.append(random.uniform(s[i],max))

    
#plotting the data for visualization purposes

#creating my own linespace
#setting
start, stop, num = 0,10,10

def linsp(start, stop, num):
    return [start + i *(stop-start)/(num-1) for i in range (num)]

x = linsp(start, stop, num)


plt.plot(x,l, color='r', linestyle=':', linewidth=1.5, marker='o')
plt.plot(x,h, color='g', linestyle='-.', linewidth=1.5, marker='*')
plt.plot(x,s, color='b', linestyle='-', linewidth=1.5, marker='s')

plt.show()



#other ways of plotting


    
```
